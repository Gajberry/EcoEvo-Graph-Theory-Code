Generate the Moran data used by `analyze_results.ipynb`. Configure `evoLudo_jar` in `jar_locations.py` before running. Existing results are not overwritten.

```python
import sys
from pathlib import Path
from config import *
sys.path.insert(0, str(next(p for p in Path(folder_loc).parents if (p / 'jar_locations.py').is_file())))
from generation import generate_arguments_string, run_simulation

from jar_locations import evoLudo_jar
```

```python
for graph in graphs:
    graph_codes = [f'r{k}' for k in ks] if graph['code'] == 'r' else [graph['code']]
    for comparison in graph['comparison_graphs']:
        graph_codes.extend([f'r{k}' for k in ks] if comparison['code'] == 'r' else [comparison['code']])
    for graph_code in dict.fromkeys(graph_codes):
        for update in graph['updates']:
            for fitness in rs:
                name = f"Moran-N-{graph['pop_size']}-geom-{graph_code}-r-{fitness}-u-{update}"
                run_simulation(results_loc, name, evoLudo_jar, generate_arguments_string({
                    'module': 'Moran',
                    'digits': 6,
                    'delay': 0,
                    'timestep': 1,
                    'popsize': graph['pop_size'],
                    'popupdate': update,
                    'geometry': graph_code,
                    'fitness': f'1,{fitness}',
                    'mutation': 0,
                    'init': ('mutant', '1,0'),
                    'samples': samples,
                    'data': 'statprob,stattimes',
                }))
```
