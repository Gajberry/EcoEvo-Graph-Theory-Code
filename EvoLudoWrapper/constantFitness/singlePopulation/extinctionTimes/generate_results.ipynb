```python
import sys
from pathlib import Path
from config import *
sys.path.insert(0, str(next(p for p in Path(folder_loc).parents if (p / 'jar_locations.py').is_file())))
from generation import generate_arguments_string, run_simulation

from jar_locations import simEMoran_jar
```

```python
graph, pop_size, d, fitness, timestep = 'M', 30, 0.5, 1, 0.001
for repeat_num in range(1000):
    name = f'CC-N-{pop_size}-geom-{graph}-d-{d}-r-{fitness}-tr-{timestep}-repeat-{repeat_num}'
    run_simulation(results_loc, name, simEMoran_jar, generate_arguments_string({
        'module': 'eMoran',
        'delay': 0,
        'timestep': timestep,
        'popsize': pop_size,
        'popupdate': 'e',
        'mutation': 0,
        'geometry': graph,
        'fitness': f'{fitness},{fitness}',
        'deathrate': d,
        'init': ('mono', '1,0'),
        'timestop': timestop,
        'timerelax': timerelax,
    }))
```
